```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
from google.colab import files
import cv2
import matplotlib.pyplot as plt

# Is code ko run karne par aik button aayega jahan se aap image upload karenge
uploaded = files.upload()

for fn in uploaded.keys():
    img_path = fn
    print(f"File uploaded successfully: {fn}")

# Image ko grayscale mein load karein
gray_img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

# Check aur display karein
if gray_img is not None:
    print("Image Shape:", gray_img.shape)
    plt.figure(figsize=(5, 5))
    plt.imshow(gray_img, cmap='gray', vmin=0, vmax=255)
    plt.title("Uploaded Grayscale Image")
    plt.axis('off')
    plt.show()
else:
    print("Error loading image!")
```

```python
# Task 1: Brightness Adjustment (β = +40 and β = -40)
bright_img = np.clip(gray_img.astype(np.float32) + 40, 0, 255).astype(np.uint8)
darker_img = np.clip(gray_img.astype(np.float32) - 40, 0, 255).astype(np.uint8)

# Nested Loop Implementation for +40 verification
h, w = gray_img.shape
loop_bright = np.zeros((h, w), dtype=np.uint8)
for i in range(h):
    for j in range(w):
        val = int(gray_img[i, j]) + 40
        loop_bright[i, j] = 255 if val > 255 else val

# Verify both NumPy and Loop results match
assert np.array_equal(bright_img, loop_bright), "Loop output does not match NumPy output!"
print("Task 1 Verification Passed: NumPy and Nested Loop results match perfectly!")

# Display Results
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(gray_img, cmap='gray', vmin=0, vmax=255)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(bright_img, cmap='gray', vmin=0, vmax=255)
plt.title('Brighter (β = +40)')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(darker_img, cmap='gray', vmin=0, vmax=255)
plt.title('Darker (β = -40)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Task 2: Linear Contrast from Zero (α = 1.5, β = 0)
linear_contrast = np.clip(gray_img.astype(np.float32) * 1.5, 0, 255).astype(np.uint8)

# Display Results
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray_img, cmap='gray', vmin=0, vmax=255)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(linear_contrast, cmap='gray', vmin=0, vmax=255)
plt.title('Linear Contrast (α = 1.5, β = 0)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Task 3: Contrast Centered at 128
contrast_15 = np.clip((gray_img.astype(np.float32) - 128) * 1.5 + 128, 0, 255).astype(np.uint8)
contrast_05 = np.clip((gray_img.astype(np.float32) - 128) * 0.5 + 128, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1), plt.imshow(gray_img, cmap='gray', vmin=0, vmax=255), plt.title('Original'), plt.axis('off')
plt.subplot(1, 3, 2), plt.imshow(contrast_15, cmap='gray', vmin=0, vmax=255), plt.title('Centered Contrast (α=1.5)'), plt.axis('off')
plt.subplot(1, 3, 3), plt.imshow(contrast_05, cmap='gray', vmin=0, vmax=255), plt.title('Centered Contrast (α=0.5)'), plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Task 4: Image Negative
negative_img = 255 - gray_img

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1), plt.imshow(gray_img, cmap='gray', vmin=0, vmax=255), plt.title('Original'), plt.axis('off')
plt.subplot(1, 2, 2), plt.imshow(negative_img, cmap='gray', vmin=0, vmax=255), plt.title('Image Negative'), plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Task 5: Gamma Correction
gamma_05 = np.clip(255.0 * ((gray_img / 255.0) ** 0.5), 0, 255).astype(np.uint8)
gamma_20 = np.clip(255.0 * ((gray_img / 255.0) ** 2.0), 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1), plt.imshow(gray_img, cmap='gray', vmin=0, vmax=255), plt.title('Original'), plt.axis('off')
plt.subplot(1, 3, 2), plt.imshow(gamma_05, cmap='gray', vmin=0, vmax=255), plt.title('Gamma (γ=0.5)'), plt.axis('off')
plt.subplot(1, 3, 3), plt.imshow(gamma_20, cmap='gray', vmin=0, vmax=255), plt.title('Gamma (γ=2.0)'), plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Task 6: Grayscale Histogram Plot
plt.figure(figsize=(6, 4))
plt.hist(gray_img.ravel(), bins=256, range=[0, 256], color='black', alpha=0.8)
plt.title("Grayscale Histogram (0-256)")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency Count")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# Task 7: Global Histogram Equalization
equalized_img = cv2.equalizeHist(gray_img)

plt.figure(figsize=(12, 8))
plt.subplot(2, 2, 1), plt.imshow(gray_img, cmap='gray', vmin=0, vmax=255), plt.title('Original Image'), plt.axis('off')
plt.subplot(2, 2, 2), plt.imshow(equalized_img, cmap='gray', vmin=0, vmax=255), plt.title('Global Equalized Image'), plt.axis('off')
plt.subplot(2, 2, 3), plt.hist(gray_img.ravel(), bins=256, range=[0, 256], color='black'), plt.title('Original Histogram')
plt.subplot(2, 2, 4), plt.hist(equalized_img.ravel(), bins=256, range=[0, 256], color='black'), plt.title('Equalized Histogram')
plt.tight_layout()
plt.show()
```

```python
# Task 8: CLAHE Implementation
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
clahe_img = clahe.apply(gray_img)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1), plt.imshow(gray_img, cmap='gray', vmin=0, vmax=255), plt.title('Original'), plt.axis('off')
plt.subplot(1, 3, 2), plt.imshow(equalized_img, cmap='gray', vmin=0, vmax=255), plt.title('Global Equalization'), plt.axis('off')
plt.subplot(1, 3, 3), plt.imshow(clahe_img, cmap='gray', vmin=0, vmax=255), plt.title('CLAHE (8x8 Grid)'), plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Task 9: Fixed Binary Thresholding (T = 127)
_, thresh_fixed = cv2.threshold(gray_img, 127, 255, cv2.THRESH_BINARY)
print("Unique pixel values in Fixed Binary Mask:", np.unique(thresh_fixed))

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1), plt.imshow(gray_img, cmap='gray', vmin=0, vmax=255), plt.title('Original'), plt.axis('off')
plt.subplot(1, 2, 2), plt.imshow(thresh_fixed, cmap='gray', vmin=0, vmax=255), plt.title('Fixed Threshold (T=127)'), plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Task 10: Otsu Thresholding
otsu_val, thresh_otsu = cv2.threshold(gray_img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
print(f"Computed Optimal Otsu Threshold Value (T*): {otsu_val}")

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1), plt.imshow(thresh_fixed, cmap='gray', vmin=0, vmax=255), plt.title('Fixed Threshold (T=127)'), plt.axis('off')
plt.subplot(1, 2, 2), plt.imshow(thresh_otsu, cmap='gray', vmin=0, vmax=255), plt.title(f'Otsu Threshold (T*={otsu_val})'), plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Task 11: Adaptive Gaussian Thresholding
thresh_adaptive = cv2.adaptiveThreshold(
    gray_img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY, 11, 2
)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1), plt.imshow(gray_img, cmap='gray', vmin=0, vmax=255), plt.title('Original'), plt.axis('off')
plt.subplot(1, 3, 2), plt.imshow(thresh_otsu, cmap='gray', vmin=0, vmax=255), plt.title('Otsu Threshold'), plt.axis('off')
plt.subplot(1, 3, 3), plt.imshow(thresh_adaptive, cmap='gray', vmin=0, vmax=255), plt.title('Adaptive Gaussian (11x11)'), plt.axis('off')
plt.tight_layout()
plt.show()
```
